## Data Analysis using Matplotlib and Seaborn

```python
data['hotel'].value_counts()
```

```python
sns.set_style(style='darkgrid')
sns.countplot(data=data,x=data['hotel'])
plt.show()
```

```python
month_data=data['arrival_date_month'].value_counts()
```

```python
month_data=pd.DataFrame(month_data).reset_index()
```

```python
month_data
```

```python
sns.set_style('white')
month_order=['Januray','February','March','April','May','June','July','August','September','October','November','December']
sns.barplot(data=month_data,x=month_data['arrival_date_month'],y=month_data['count'],order=month_order)
plt.xticks(rotation=45)
plt.show()
```

```python
cancel=data['is_canceled'].value_counts()
```

```python
cancel
```

```python
plt.pie(cancel.values,labels=cancel.index,autopct='%.f',shadow=True)
plt.show()
```

```python
country_booking=data['country'].value_counts().sort_values(ascending=False).head(10)
```

```python
sns.barplot(data=country_booking.to_frame(),x=country_booking.index,y=country_booking.values)
plt.show()
```

```python
cust_type=data['customer_type'].value_counts()
```

```python
cust_type
```

```python
plt.pie(cust_type.values,labels=cust_type.index,autopct='%.2f',shadow=True)
plt.show()
```

```python
arrival_week=data['arrival_date_week_number'].value_counts().sort_index()
```

```python
sns.lineplot(x=arrival_week.index,y=arrival_week.values)
plt.show()
```

```python
cancelled=pd.crosstab(data['customer_type'],data['is_canceled'])
```

```python
sns.heatmap(cancelled,cmap='coolwarm')
plt.show()
```

```python
sns.countplot(data=data,x='hotel',hue='is_canceled')
plt.show()
```

```python
market=pd.crosstab(data['market_segment'],data['is_canceled'])
```

```python
market
```

```python
sns.heatmap(market,cmap='viridis')
plt.show()
```

```python
data['is_repeated_guest'].value_counts()
```

```python
repeated=(pd.crosstab(data['hotel'],data['is_repeated_guest'],normalize=True)*100).round(2)
```

```python
repeated
```

```python
sns.countplot(data=data,x='hotel',hue='is_repeated_guest')
plt.show()
```

```python
sns.catplot(data=data,x='is_repeated_guest',hue='is_canceled',col='hotel',kind='count')
plt.show()
```

```python
channel= data['distribution_channel'].value_counts()
```

```python
channel
```

```python
sns.kdeplot(data=data,x='adr',hue='hotel')
plt.show()
```

```python
print(data['distribution_channel'].unique())
print(data['reserved_room_type'].unique())
```

```python
grp=data.groupby(['distribution_channel','reserved_room_type']).size().reset_index(name='count')
```

```python
grp
```

```python
sns.catplot(data=grp,x='reserved_room_type',y='count',col='distribution_channel',kind='point')
plt.show()
```

```python
sns.boxplot(data=data,y=data['lead_time'],hue=data['is_canceled'],palette='viridis',fliersize=2)
plt.xlabel('is_canceled')
plt.show()
```
